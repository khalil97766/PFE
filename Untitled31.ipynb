```python
import cv2
import os
import numpy as np
import time
import geopandas as gpd
from shapely.geometry import Polygon, MultiPolygon # Import the geometry types

class DynamicObstacle:
    def __init__(self, x, y, x_end, speed):
        self.pos = np.array([x, y], dtype=float)
        self.start_pos = np.array([x, y], dtype=float)
        self.end_pos = np.array([x_end, y], dtype=float)
        self.speed = speed
        self.direction = 1

    def update(self):
        self.pos[0] += self.speed * self.direction
        if self.pos[0] >= self.end_pos[0] or self.pos[0] <= self.start_pos[0]:
            self.direction *= -1

def initialize_robot_hardware():
    try:
        cap = cv2.VideoCapture(0)
        if not cap.isOpened(): cap = None
    except Exception:
        cap = None
    return {'camera': cap}

def get_live_sensor_data(hardware):
    if hardware['camera'] is None: return None
    ret, frame = hardware['camera'].read()
    return frame if ret else None

def get_current_robot_position(global_path):
    if not hasattr(get_current_robot_position, "index"):
        get_current_robot_position.index = 0
    current_pos = global_path[get_current_robot_position.index]
    if get_current_robot_position.index < len(global_path) - 1:
        get_current_robot_position.index += 1
    return tuple(current_pos)

def actuate_motors(direction_vector):
    if np.linalg.norm(direction_vector) < 1e-6:
        pass
    else:
        pass

def shutdown_robot(hardware):
    if hardware['camera'] is not None:
        hardware['camera'].release()

# ===================================================================
# THIS FUNCTION IS NOW FIXED
# ===================================================================
def create_static_costmap(width, height, buildings_gdf, global_path_coords, settings):
    costmap = np.full((height, width), settings['cost_sidewalk'], dtype=np.uint8)
    path_min = np.min(global_path_coords, axis=0)
    path_max = np.max(global_path_coords, axis=0)
    map_origin_world = path_min - 0.001
    map_span_world = (path_max - path_min) + 0.002
    scale_x = width / map_span_world[0] if map_span_world[0] != 0 else 1
    scale_y = height / map_span_world[1] if map_span_world[1] != 0 else 1
    transform = {'origin': map_origin_world, 'scale_x': scale_x, 'scale_y': scale_y}
    
    for geom in buildings_gdf.geometry:
        if geom is None: continue

        # --- THIS IS THE KEY FIX ---
        # We now check the geometry type to handle both simple and complex polygons.
        
        # If the geometry is a simple polygon, draw it directly.
        if isinstance(geom, Polygon):
            polygons_to_draw = [geom]
        # If it's a MultiPolygon, create a list of all the simple polygons it contains.
        elif isinstance(geom, MultiPolygon):
            polygons_to_draw = list(geom.geoms)
        else:
            continue # Ignore other types like points or lines in the building file

        # Loop through the list of polygons (will be 1 for a Polygon, or more for a MultiPolygon)
        for poly in polygons_to_draw:
            pixel_coords = []
            for lon, lat in poly.exterior.coords:
                px = int((lon - transform['origin'][0]) * transform['scale_x'])
                py = int((lat - transform['origin'][1]) * transform['scale_y'])
                pixel_coords.append([px, height - py])
            cv2.fillPoly(costmap, [np.array(pixel_coords)], settings['cost_obstacle'])

    return costmap, transform

def world_to_map_coords(world_pos, map_height, transform_info):
    mx = int((world_pos[0] - transform_info['origin'][0]) * transform_info['scale_x'])
    my = int((world_pos[1] - transform_info['origin'][1]) * transform_info['scale_y'])
    return (mx, map_height - my)

def get_local_goal(robot_pos, global_path, lookahead_distance):
    robot_pos_arr = np.array(robot_pos)
    path_points = np.array(global_path)
    distances = np.linalg.norm(path_points - robot_pos_arr, axis=1)
    closest_idx = np.argmin(distances)
    lookahead_idx = min(closest_idx + lookahead_distance, len(global_path) - 1)
    return global_path[lookahead_idx]

def calculate_apf_direction(costmap, robot_map_pos, local_goal_map_pos, settings, dynamic_obstacles_map_pos=[]):
    k_att, k_rep, rep_radius = settings['k_att'], settings['k_rep'], settings['rep_radius']
    robot_pos_arr = np.array(robot_map_pos, dtype=float)
    goal_pos_arr = np.array(local_goal_map_pos, dtype=float)
    
    attractive_force = k_att * (goal_pos_arr - robot_pos_arr)
    repulsive_force = np.zeros(2, dtype=float)
    
    obs_y, obs_x = np.where(costmap >= settings['cost_obstacle'])
    if obs_x.size > 0:
        obstacle_positions = np.stack([obs_x, obs_y], axis=1)
        distances_to_obs = np.linalg.norm(obstacle_positions - robot_pos_arr, axis=1)
        close_obstacles = obstacle_positions[distances_to_obs < rep_radius]
        for obs_pos in close_obstacles:
            vec_to_obs = robot_pos_arr - obs_pos
            dist = np.linalg.norm(vec_to_obs)
            if dist > 0.1:
                force_mag = k_rep * (1.0/dist - 1.0/rep_radius) / (dist**2)
                repulsive_force += force_mag * (vec_to_obs / dist)

    for obs_pos in dynamic_obstacles_map_pos:
        vec_to_obs = robot_pos_arr - np.array(obs_pos)
        dist = np.linalg.norm(vec_to_obs)
        if dist < rep_radius and dist > 0.1:
            force_mag = k_rep * (1.0/dist - 1.0/rep_radius) / (dist**2)
            repulsive_force += force_mag * (vec_to_obs / dist)

    net_force = attractive_force + repulsive_force
    norm = np.linalg.norm(net_force)
    return net_force / norm if norm > 1e-6 else np.array([0, 0])

def main():
    try:
        global_path = np.load('global_path.npy').tolist()
        buildings_gdf = gpd.read_file('bejaia_buildings.geojson')
    except FileNotFoundError:
        return
        
    settings = {
        'map_width': 800, 'map_height': 800, 'cost_sidewalk': 1, 'cost_obstacle': 255,
        'local_path_lookahead': 10, 'k_att': 1.0, 'k_rep': 250.0, 'rep_radius': 50.0,
        'loop_frequency': 10
    }
    
    dynamic_obstacles = [DynamicObstacle(x=5.068, y=36.751, x_end=5.072, speed=0.00005)]
    
    GLOBAL_END = tuple(global_path[-1])
    static_costmap, transform = create_static_costmap(settings['map_width'], settings['map_height'], buildings_gdf, global_path, settings)
    viz_map_base = cv2.cvtColor(static_costmap, cv2.COLOR_GRAY2BGR)

    hardware = initialize_robot_hardware()
    
    try:
        while True:
            start_time = time.time()
            frame = get_live_sensor_data(hardware)
            cam_view = frame if frame is not None else np.zeros((480, 640, 3), dtype=np.uint8)
            
            for obs in dynamic_obstacles: obs.update()
            dynamic_obs_map = [world_to_map_coords(obs.pos, settings['map_height'], transform) for obs in dynamic_obstacles]

            robot_world_pos = get_current_robot_position(global_path)
            
            dist_to_end = np.linalg.norm(np.array(robot_world_pos) - np.array(GLOBAL_END))
            if dist_to_end < 0.0001:
                direction = np.array([0, 0])
            else:
                local_goal_world = get_local_goal(robot_world_pos, global_path, settings['local_path_lookahead'])
                robot_map_pos = world_to_map_coords(robot_world_pos, settings['map_height'], transform)
                local_goal_map = world_to_map_coords(local_goal_world, settings['map_height'], transform)
                direction = calculate_apf_direction(static_costmap, robot_map_pos, local_goal_map, settings, dynamic_obs_map)
            
            actuate_motors(direction)

            viz_map = viz_map_base.copy()
            path_map_coords = [world_to_map_coords(p, settings['map_height'], transform) for p in global_path]
            for i in range(len(path_map_coords) - 1):
                cv2.line(viz_map, path_map_coords[i], path_map_coords[i+1], (255, 0, 0), 2)
            robot_map_viz_pos = world_to_map_coords(robot_world_pos, settings['map_height'], transform)
            cv2.circle(viz_map, robot_map_viz_pos, 8, (0, 255, 0), -1)
            for obs_pos in dynamic_obs_map:
                cv2.circle(viz_map, obs_pos, 8, (0, 165, 255), -1)
            
            cv2.imshow('Bejaia Navigation Map', viz_map)
            cv2.imshow('Robot Camera View', cam_view)
            
            if cv2.waitKey(1) & 0xFF == ord('q'): break
            time.sleep(max(0, (1.0/settings['loop_frequency']) - (time.time() - start_time)))

    finally:
        shutdown_robot(hardware)
        cv2.destroyAllWindows()

if __name__ == '__main__':
    main()
```

```python
# ===================================================================
# المكتبات المطلوبة
# ===================================================================
import cv2
import os
import numpy as np
import time
import geopandas as gpd
from shapely.geometry import Polygon, MultiPolygon, Point, LineString
import pynmea2
import socket # مكتبة للاتصالات الشبكية

# ===================================================================
# الدوال الأساسية للروبوت (بدون تغيير)
# ===================================================================

def actuate_motors(direction_vector):
    """دالة وهمية لمحاكاة تشغيل المحركات."""
    # في روبوت حقيقي، هذا الكود سيرسل أوامر إلى المحركات.
    # print(f"INFO: Actuating motors with direction: {direction_vector}")
    pass

def create_static_costmap(width, height, buildings_gdf, global_path_coords, settings):
    costmap = np.full((height, width), settings['cost_sidewalk'], dtype=np.uint8)
    path_min = np.min(global_path_coords, axis=0)
    path_max = np.max(global_path_coords, axis=0)
    map_origin_world = path_min - 0.001
    map_span_world = (path_max - path_min) + 0.002
    scale_x = width / map_span_world[0] if map_span_world[0] != 0 else 1
    scale_y = height / map_span_world[1] if map_span_world[1] != 0 else 1
    transform = {'origin': map_origin_world, 'scale_x': scale_x, 'scale_y': scale_y}
    for geom in buildings_gdf.geometry:
        polygons_to_draw = []
        if isinstance(geom, Polygon): polygons_to_draw = [geom]
        elif isinstance(geom, MultiPolygon): polygons_to_draw = list(geom.geoms)
        for poly in polygons_to_draw:
            pixel_coords = []
            for lon, lat in poly.exterior.coords:
                px = int((lon - transform['origin'][0]) * transform['scale_x'])
                py = int((lat - transform['origin'][1]) * transform['scale_y'])
                pixel_coords.append([px, height - py])
            cv2.fillPoly(costmap, [np.array(pixel_coords)], settings['cost_obstacle'])
    return costmap, transform

def find_closest_point_on_path(robot_pos, global_path):
    robot_point = Point(robot_pos)
    path_line = LineString(global_path)
    return path_line.interpolate(path_line.project(robot_point))
    
def get_local_goal(robot_pos, global_path, lookahead_distance):
    robot_pos_arr = np.array(robot_pos)
    path_points = np.array(global_path)
    distances = np.linalg.norm(path_points - robot_pos_arr, axis=1)
    closest_idx = np.argmin(distances)
    lookahead_idx = min(closest_idx + lookahead_distance, len(global_path) - 1)
    return global_path[lookahead_idx]

def calculate_apf_direction(robot_map_pos, local_goal_map_pos, settings):
    k_att = settings['k_att']
    robot_pos_arr = np.array(robot_map_pos, dtype=float)
    goal_pos_arr = np.array(local_goal_map_pos, dtype=float)
    attractive_force = k_att * (goal_pos_arr - robot_pos_arr)
    norm = np.linalg.norm(attractive_force)
    return attractive_force / norm if norm > 1e-6 else np.array([0, 0])

def world_to_map_coords(world_pos, map_height, transform_info):
    mx = int((world_pos[0] - transform_info['origin'][0]) * transform_info['scale_x'])
    my = int((world_pos[1] - transform_info['origin'][1]) * transform_info['scale_y'])
    return (mx, map_height - my)

# ===================================================================
# دوال جديدة للتعامل مع مكونات الهاتف (GPS + كاميرا)
# ===================================================================

def initialize_phone_gps(ip, port):
    """تهيئة اتصال شبكي (socket) لاستقبال بيانات الـ GPS من الهاتف."""
    try:
        sock = socket.socket(socket.AF_INET, socket.SOCK_STREAM)
        sock.settimeout(2) # تحديد مهلة زمنية لتجنب التوقف إلى أجل غير مسمى
        server_address = (ip, port)
        print(f"INFO: جاري الاتصال بـ GPS الهاتف على العنوان {ip}:{port}...")
        sock.connect(server_address)
        sock.settimeout(None) # إزالة المهلة الزمنية بعد الاتصال
        print("INFO: تم الاتصال بنجاح بـ GPS الهاتف.")
        return sock
    except Exception as e:
        print(f"خطأ فادح: لا يمكن الاتصال بالهاتف. هل تطبيق البث يعمل؟ الخطأ: {e}")
        return None

def get_gps_reading_from_phone(sock):
    """قراءة وتحليل سطر بيانات GPS بصيغة NMEA من الاتصال الشبكي."""
    if sock is None:
        return None
    try:
        # استقبال البيانات وفك تشفيرها. حجم المخزن المؤقت 4096 كافٍ بشكل عام.
        data = sock.recv(4096).decode('ascii', errors='replace')
        # قد تحتوي البيانات على عدة جمل NMEA، لذا نفصلها حسب السطر
        for line in data.strip().split('\n'):
            if line.startswith('$GPGGA'):
                msg = pynmea2.parse(line)
                # التحقق مما إذا كانت الجملة تحتوي على إشارة GPS صالحة
                if getattr(msg, 'is_valid', False) and msg.longitude != 0.0 and msg.latitude != 0.0:
                    return (msg.longitude, msg.latitude)
    except BlockingIOError:
        # لا توجد بيانات للقراءة، هذا طبيعي
        pass
    except Exception as e:
        print(f"تحذير: حدث خطأ أثناء قراءة بيانات الـ GPS: {e}")
        return None # إرجاع None في حالة فقدان الاتصال
    return None

def initialize_phone_camera(url):
    """تهيئة التقاط الفيديو من رابط كاميرا الهاتف."""
    cap = cv2.VideoCapture(url)
    if not cap.isOpened():
        print(f"خطأ فادح: لا يمكن فتح بث الكاميرا من الرابط: {url}")
        return None
    print(f"INFO: تم الاتصال بنجاح بكاميرا الهاتف.")
    return cap

def shutdown_robot(gps_socket, camera_capture):
    """إغلاق الاتصالات الشبكية ونوافذ OpenCV بشكل صحيح."""
    if gps_socket is not None:
        gps_socket.close()
        print("INFO: تم إغلاق اتصال الـ GPS.")
    if camera_capture is not None:
        camera_capture.release()
        print("INFO: تم تحرير التقاط الكاميرا.")
    cv2.destroyAllWindows()
    print("INFO: تم إغلاق جميع النوافذ.")

# ===================================================================
# الدالة الرئيسية
# ===================================================================
def main():
    # --- إعدادات يجب عليك ضبطها ---
    # استبدل هذه القيم بالقيم الخاصة بهاتفك
    PHONE_IP = "192.168.1.12"  # ضع عنوان IP الخاص بهاتفك هنا
    GPS_PORT = 50000           # ضع رقم المنفذ (Port) الخاص بتطبيق GPS هنا
    CAMERA_URL = "http://192.168.1.12:8080/video" # ضع رابط URL الخاص بتطبيق IP Webcam هنا

    try:
        global_path = np.load('global_path.npy').tolist()
        buildings_gdf = gpd.read_file('bejaia_buildings.geojson')
    except FileNotFoundError as e:
        print(f"خطأ فادح: الملف '{e.filename}' غير موجود. تأكد من أن ملفات البيانات موجودة في نفس المجلد.")
        return
        
    settings = {
        'map_width': 800, 'map_height': 800, 'cost_sidewalk': 1, 'cost_obstacle': 255,
        'local_path_lookahead': 10, # تمت زيادتها لتنقل أكثر سلاسة
        'k_att': 1.0, 
        'loop_frequency': 5 # Hz
    }
    
    GLOBAL_END = tuple(global_path[-1])
    static_costmap, transform = create_static_costmap(settings['map_width'], settings['map_height'], buildings_gdf, global_path, settings)
    viz_map_base = cv2.cvtColor(static_costmap, cv2.COLOR_GRAY2BGR)

    # تهيئة مكونات الهاتف (GPS وكاميرا عبر Wi-Fi)
    gps_socket = initialize_phone_gps(PHONE_IP, GPS_PORT)
    camera_capture = initialize_phone_camera(CAMERA_URL)
    
    # إذا فشل اتصال GPS، لا نستمر
    if gps_socket is None:
        print("إيقاف البرنامج لعدم التمكن من إنشاء اتصال GPS.")
        return

    try:
        while True:
            start_time = time.time()
            
            # --- قراءة بيانات الحساسات ---
            robot_gps_pos = get_gps_reading_from_phone(gps_socket)
            
            if camera_capture is not None:
                ret, camera_frame = camera_capture.read()
            else:
                ret = False

            # --- منطق الملاحة ---
            if robot_gps_pos is None:
                print("...في انتظار الحصول على إشارة GPS صالحة")
                time.sleep(1)
                continue

            dist_to_end = np.linalg.norm(np.array(robot_gps_pos) - np.array(GLOBAL_END))
            if dist_to_end < 0.0002: # عتبة المسافة لاعتبار الهدف قد تم الوصول إليه
                print("INFO: تم الوصول إلى الهدف!")
                direction = np.array([0, 0])
                actuate_motors(direction)
                break # الخروج من الحلقة
            else:
                local_goal_world = get_local_goal(robot_gps_pos, global_path, settings['local_path_lookahead'])
                robot_map_pos = world_to_map_coords(robot_gps_pos, settings['map_height'], transform)
                local_goal_map = world_to_map_coords(local_goal_world, settings['map_height'], transform)
                direction = calculate_apf_direction(robot_map_pos, local_goal_map, settings)
                actuate_motors(direction)

            # --- العرض المرئي ---
            viz_map = viz_map_base.copy()
            # رسم المسار العام
            path_map_coords = [world_to_map_coords(p, settings['map_height'], transform) for p in global_path]
            cv2.polylines(viz_map, [np.array(path_map_coords)], isClosed=False, color=(255, 0, 0), thickness=2)
            
            # رسم موقع الروبوت (دائرة خضراء)
            robot_gps_map_pos = world_to_map_coords(robot_gps_pos, settings['map_height'], transform)
            cv2.circle(viz_map, robot_gps_map_pos, 8, (0, 255, 0), -1)

            # رسم خط إلى أقرب نقطة على المسار
            closest_point_world = find_closest_point_on_path(robot_gps_pos, global_path)
            closest_point_map = world_to_map_coords(closest_point_world.coords[0], settings['map_height'], transform)
            cv2.line(viz_map, robot_gps_map_pos, closest_point_map, (0, 255, 255), 1)

            # عرض خريطة الملاحة
            cv2.imshow('ملاحة الروبوت في الوقت الحقيقي', viz_map)
            
            # عرض بث الكاميرا إذا كان متاحًا
            if ret:
                cv2.imshow('بث كاميرا الهاتف', camera_frame)
            
            # 'q' للخروج اضغط على حرف
            if cv2.waitKey(1) & 0xFF == ord('q'):
                actuate_motors(np.array([0,0]))
                break
            
            # التحكم في تردد الحلقة
            time.sleep(max(0, (1.0/settings['loop_frequency']) - (time.time() - start_time)))

    finally:
        # التأكد من الإغلاق الصحيح
        shutdown_robot(gps_socket, camera_capture)

if __name__ == '__main__':
    main()
```
