```python
import cv2
import os
import numpy as np
import time
import geopandas as gpd
from shapely.geometry import Polygon, MultiPolygon, Point, LineString
import pynmea2
import socket

# --- SLAM/Sensor Fusion Additions ---
# For a more complete SLAM, you'd use a dedicated library or a full EKF/Particle Filter implementation.
# This is a very basic visual odometry and GPS fusion placeholder.

class RobotState:
    def __init__(self, initial_lon, initial_lat):
        self.lon = initial_lon
        self.lat = initial_lat
        self.yaw = 0.0 # Orientation in radians
        self.velocity_x = 0.0
        self.velocity_y = 0.0
        self.P = np.eye(5) * 0.1 # Covariance matrix for EKF

    def predict(self, dt, angular_velocity_z=0, linear_velocity_x=0, linear_velocity_y=0):
        # Basic motion model (for demonstration)
        # In a real system, you'd use IMU data here.
        self.lon += self.velocity_x * dt
        self.lat += self.velocity_y * dt
        self.yaw += angular_velocity_z * dt
        # Propagate covariance (requires Jacobian of motion model)
        # self.P = F @ self.P @ F.T + Q (where F is Jacobian, Q is process noise)

    def update_gps(self, new_lon, new_lat, gps_noise_covariance=0.1):
        # Very simple GPS update (like a direct measurement update)
        # In an EKF, this would involve computing Kalman gain.
        self.lon = (self.lon * self.P[0,0] + new_lon * gps_noise_covariance) / (self.P[0,0] + gps_noise_covariance)
        self.lat = (self.lat * self.P[1,1] + new_lat * gps_noise_covariance) / (self.P[1,1] + gps_noise_covariance)
        # Update covariance (requires more complex EKF math)

    def update_visual_odometry(self, delta_x, delta_y, delta_yaw, vo_noise_covariance=0.01):
        # Update based on relative motion from visual odometry
        self.lon += delta_x # These would be transformed to world coords
        self.lat += delta_y
        self.yaw += delta_yaw
        # Update covariance


def initialize_feature_detector():
    """Initializes ORB detector for feature tracking."""
    # You might want SIFT/SURF for more robustness, but ORB is faster.
    # For a real SLAM system, you'd use a feature-based Visual Odometry library.
    return cv2.ORB_create()

def track_features_and_estimate_pose(prev_frame_gray, current_frame_gray, prev_keypoints, prev_descriptors, orb_detector):
    """
    Tracks features between frames and estimates relative pose.
    This is a highly simplified placeholder. A real VO system is complex.
    """
    current_keypoints, current_descriptors = orb_detector.detectAndCompute(current_frame_gray, None)

    if prev_descriptors is None or current_descriptors is None or len(prev_descriptors) == 0 or len(current_descriptors) == 0:
        return None, None, current_keypoints, current_descriptors

    # Use Brute-Force Matcher
    bf = cv2.BFMatcher(cv2.NORM_HAMMING, crossCheck=True)
    matches = bf.match(prev_descriptors, current_descriptors)
    matches = sorted(matches, key=lambda x: x.distance)

    # Select good matches
    good_matches = matches[:50] # Take top N matches

    if len(good_matches) < 8: # Need at least 8 points for fundamental/essential matrix
        return None, None, current_keypoints, current_descriptors

    src_pts = np.float32([prev_keypoints[m.queryIdx].pt for m in good_matches]).reshape(-1, 1, 2)
    dst_pts = np.float32([current_keypoints[m.trainIdx].pt for m in good_matches]).reshape(-1, 1, 2)

    # Estimate fundamental matrix and essential matrix (if camera intrinsics are known)
    # F, mask = cv2.findFundamentalMat(src_pts, dst_pts, cv2.FM_RANSAC)

    # For relative pose estimation, you need camera intrinsics and potentially 3D points
    # This example is highly simplified. You'd typically use solvePnP with 3D-2D correspondences
    # or estimate Essential Matrix for relative pose if you have intrinsic parameters.
    # For a basic Visual Odometry, you might approximate rotation and translation.

    # Placeholder for estimated delta_x, delta_y, delta_yaw
    # In a real scenario, this would come from camera pose estimation (e.g., cv2.recoverPose)
    # after estimating the Essential Matrix and triangulating points.
    delta_x = 0.0 # Placeholder
    delta_y = 0.0 # Placeholder
    delta_yaw = 0.0 # Placeholder

    return (delta_x, delta_y, delta_yaw), good_matches, current_keypoints, current_descriptors

# --- End SLAM/Sensor Fusion Additions ---

# Your existing functions remain the same
def actuate_motors(direction_vector):
    # This would send commands to motors
    # For now, print the direction
    print(f"Actuating motors with direction: {direction_vector}")
    pass

def create_static_costmap(width, height, buildings_gdf, global_path_coords, settings):
    costmap = np.full((height, width), settings['cost_sidewalk'], dtype=np.uint8)
    path_min = np.min(global_path_coords, axis=0)
    path_max = np.max(global_path_coords, axis=0)
    map_origin_world = path_min - 0.001
    map_span_world = (path_max - path_min) + 0.002
    scale_x = width / map_span_world[0] if map_span_world[0] != 0 else 1
    scale_y = height / map_span_world[1] if map_span_world[1] != 0 else 1
    transform = {'origin': map_origin_world, 'scale_x': scale_x, 'scale_y': scale_y}

    for geom in buildings_gdf.geometry:
        polygons_to_draw = []
        if isinstance(geom, Polygon): polygons_to_draw = [geom]
        elif isinstance(geom, MultiPolygon): polygons_to_draw = list(geom.geoms)
        for poly in polygons_to_draw:
            pixel_coords = []
            for lon, lat in poly.exterior.coords:
                px = int((lon - transform['origin'][0]) * transform['scale_x'])
                py = int((lat - transform['origin'][1]) * transform['scale_y'])
                pixel_coords.append([px, height - py])
            cv2.fillPoly(costmap, [np.array(pixel_coords)], settings['cost_obstacle'])
    return costmap, transform


def find_closest_point_on_path(robot_pos, global_path):
    robot_point = Point(robot_pos)
    path_line = LineString(global_path)
    return path_line.interpolate(path_line.project(robot_point))

def get_local_goal(robot_pos, global_path, lookahead_distance):
    robot_pos_arr = np.array(robot_pos)
    path_points = np.array(global_path)
    distances = np.linalg.norm(path_points - robot_pos_arr, axis=1)
    closest_idx = np.argmin(distances)
    # Ensure lookahead_distance doesn't go out of bounds relative to the current position in path
    effective_lookahead_idx = min(closest_idx + lookahead_distance, len(global_path) - 1)
    return global_path[effective_lookahead_idx]

def calculate_apf_direction(robot_map_pos, local_goal_map_pos, settings):
    k_att = settings['k_att']
    robot_pos_arr = np.array(robot_map_pos, dtype=float)
    goal_pos_arr = np.array(local_goal_map_pos, dtype=float)
    attractive_force = k_att * (goal_pos_arr - robot_pos_arr)
    norm = np.linalg.norm(attractive_force)
    return attractive_force / norm if norm > 1e-6 else np.array([0, 0])

def world_to_map_coords(world_pos, map_height, transform_info):
    mx = int((world_pos[0] - transform_info['origin'][0]) * transform_info['scale_x'])
    my = int((world_pos[1] - transform_info['origin'][1]) * transform_info['scale_y'])
    return (mx, map_height - my)

def connect_to_phone_gps(ip, port):
    """الاتصال كتطبيق عميل إلى خادم TCP في الهاتف."""
    try:
        sock = socket.socket(socket.AF_INET, socket.SOCK_STREAM)
        sock.connect((ip, port))
        print(f" تم الاتصال بتطبيق GPS على {ip}:{port}")
        return sock
    except Exception as e:
        print(f" فشل الاتصال بالهاتف: {e}")
        return None

def get_gps_from_phone(sock):
    if sock is None:
        return None
    try:
        data = sock.recv(1024).decode('ascii', errors='replace')
        for line in data.strip().split('\n'):
            if line.startswith('$GPGGA'):
                msg = pynmea2.parse(line)
                # Ensure the message has valid fix quality (1 for GPS fix)
                if getattr(msg, 'is_valid', False) and msg.longitude and msg.latitude and msg.gps_qual == 1:
                    return (msg.longitude, msg.latitude)
    except Exception:
        pass
    return None


def initialize_phone_camera(CAMERA_URL):
    cap = cv2.VideoCapture(CAMERA_URL)
    if not cap.isOpened():
        print(" فشل الاتصال بالكاميرا")
        return None
    print(" تم الاتصال بالكاميرا")
    return cap

def shutdown(gps_socket, camera_capture):
    if gps_socket: gps_socket.close()
    if camera_capture: camera_capture.release()
    cv2.destroyAllWindows()
    print(" تم إغلاق النظام بنجاح")


def main():
    PHONE_IP = "192.168.1.12" # Make sure this is your phone's IP
    GPS_PORT = 50000
    CAMERA_URL = "http://10.198.188.144:8080/video" # Double check this URL, often it's just http://IP:PORT/video

    try:
        global_path = np.load('global_path.npy').tolist()
        buildings_gdf = gpd.read_file('bejaia_buildings.geojson')
    except FileNotFoundError as e:
        print(f" الملف مفقود: {e.filename}")
        return

    settings = {
        'map_width': 800, 'map_height': 800,
        'cost_sidewalk': 1, 'cost_obstacle': 255,
        'local_path_lookahead': 10, 'k_att': 1.0, 'loop_frequency': 5,
        'apf_repulsive_gain': 1000.0, # Example parameter for APF tuning
        'apf_obstacle_radius': 0.1 # Example parameter for APF tuning
    }

    GLOBAL_END = tuple(global_path[-1])
    costmap, transform = create_static_costmap(settings['map_width'], settings['map_height'], buildings_gdf, global_path, settings)
    viz_base = cv2.cvtColor(costmap, cv2.COLOR_GRAY2BGR)

    gps_socket = connect_to_phone_gps(PHONE_IP, GPS_PORT)
    camera_capture = initialize_phone_camera(CAMERA_URL)

    if gps_socket is None:
        shutdown(None, camera_capture)
        return

    # --- SLAM/Sensor Fusion Initialization ---
    robot_state = None # Will be initialized with first GPS reading
    orb_detector = initialize_feature_detector()
    prev_frame_gray = None
    prev_keypoints = None
    prev_descriptors = None
    last_loop_time = time.time()
    # --- End SLAM Initialization ---

    try:
        while True:
            start_loop_time = time.time()
            dt = start_loop_time - last_loop_time
            last_loop_time = start_loop_time

            robot_gps_raw = get_gps_from_phone(gps_socket)
            current_frame = None
            ret = False
            if camera_capture is not None:
                ret, frame = camera_capture.read()
                if ret:
                    current_frame = frame
                    current_frame_gray = cv2.cvtColor(current_frame, cv2.COLOR_BGR2GRAY)
                else:
                    print("Could not read frame from camera.")

            # --- SLAM/Sensor Fusion Loop ---
            if robot_gps_raw is not None and robot_state is None:
                robot_state = RobotState(robot_gps_raw[0], robot_gps_raw[1])
                print(f"Robot state initialized at: {robot_state.lon}, {robot_state.lat}")

            if robot_state:
                # 1. Prediction step (if IMU or motor commands were available)
                # robot_state.predict(dt, angular_velocity_z, linear_velocity_x, linear_velocity_y)

                # 2. Update with GPS data
                if robot_gps_raw:
                    robot_state.update_gps(robot_gps_raw[0], robot_gps_raw[1])

                # 3. Update with Visual Odometry
                if current_frame is not None and prev_frame_gray is not None:
                    # You'd need a more robust VO implementation here.
                    # This is just a placeholder to show the integration point.
                    # Estimated pose change (delta_x, delta_y, delta_yaw) from VO
                    vo_pose_change, matches, current_keypoints, current_descriptors = track_features_and_estimate_pose(
                        prev_frame_gray, current_frame_gray, prev_keypoints, prev_descriptors, orb_detector
                    )
                    if vo_pose_change:
                        robot_state.update_visual_odometry(vo_pose_change[0], vo_pose_change[1], vo_pose_change[2])

                    prev_frame_gray = current_frame_gray
                    prev_keypoints = current_keypoints
                    prev_descriptors = current_descriptors
                elif current_frame is not None: # First frame
                    prev_frame_gray = current_frame_gray
                    prev_keypoints, prev_descriptors = orb_detector.detectAndCompute(prev_frame_gray, None)


                # Use the SLAM-estimated position for navigation
                robot_pos_for_nav = (robot_state.lon, robot_state.lat)
            else:
                # Fallback to raw GPS if SLAM not initialized or failed
                robot_pos_for_nav = robot_gps_raw
                print("Waiting for initial GPS fix to start SLAM...")
            # --- End SLAM/Sensor Fusion Loop ---

            if robot_pos_for_nav is None:
                print("في انتظار موقع الروبوت (GPS أو SLAM)...")
                time.sleep(1)
                continue

            dist_to_end = np.linalg.norm(np.array(robot_pos_for_nav) - np.array(GLOBAL_END))
            if dist_to_end < 0.0002:
                print("🎯 تم الوصول إلى الهدف!")
                actuate_motors(np.array([0,0]))
                break

            local_goal = get_local_goal(robot_pos_for_nav, global_path, settings['local_path_lookahead'])
            robot_map = world_to_map_coords(robot_pos_for_nav, settings['map_height'], transform)
            goal_map = world_to_map_coords(local_goal, settings['map_height'], transform)
            
            # Add dynamic obstacle avoidance based on camera (conceptual)
            # This would require image processing to detect obstacles and add repulsive forces
            # dynamic_repulsive_force = calculate_dynamic_repulsive_force(current_frame, robot_map, settings)
            direction = calculate_apf_direction(robot_map, goal_map, settings)
            # direction = direction + dynamic_repulsive_force # Combine forces
            
            actuate_motors(direction)

            # Visualization
            viz = viz_base.copy()
            path_map = [world_to_map_coords(p, settings['map_height'], transform) for p in global_path]
            cv2.polylines(viz, [np.array(path_map)], False, (255,0,0), 2)
            cv2.circle(viz, robot_map, 8, (0,255,0), -1) # Green circle for robot
            cv2.circle(viz, goal_map, 6, (0,255,255), -1) # Yellow circle for local goal

            # Optional: Visualize features if using visual odometry
            if ret and current_frame is not None and prev_frame_gray is not None and 'matches' in locals():
                img_matches = cv2.drawMatches(prev_frame_gray, prev_keypoints, current_frame_gray, current_keypoints, matches, None,
                                              flags=cv2.DrawMatchesFlags_NOT_DRAW_SINGLE_POINTS)
                # cv2.imshow('ORB Matches', img_matches) # Uncomment to see matches

            cv2.imshow('Robot Navigation', viz)
            if ret and current_frame is not None:
                cv2.imshow('Phone Camera', current_frame)
            if cv2.waitKey(1) & 0xFF == ord('q'):
                actuate_motors(np.array([0,0]))
                break

            # Maintain loop frequency
            time_spent = time.time() - start_loop_time
            sleep_time = max(0, (1.0/settings['loop_frequency']) - time_spent)
            time.sleep(sleep_time)
    finally:
        shutdown(gps_socket, camera_capture)

if __name__ == '__main__':
    main()
    
```
