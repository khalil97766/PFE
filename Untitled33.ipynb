```python
import socket
import cv2

# ===============================
# إعدادات GPS (TCP)
# ===============================
GPS_IP = "192.168.1.12"   # ← غيّرها بعنوان الهاتف
GPS_PORT = 50000          # ← نفس المنفذ في تطبيق GPS TCP Server

# ===============================
# إعدادات الكاميرا (IP Webcam)
# ===============================
CAMERA_URL = "http://10.198.188.144:8080/video"  # ← غيّرها بعنوان هاتفك في تطبيق IP Webcam


def test_gps_connection():
    print("\n اختبار اتصال GPS TCP...")
    try:
        s = socket.socket(socket.AF_INET, socket.SOCK_STREAM)
        s.settimeout(5)
        s.connect((GPS_IP, GPS_PORT))
        print(" تم الاتصال بخادم GPS بنجاح!")
        s.close()
        return True
    except Exception as e:
        print(" فشل الاتصال بـ GPS:", e)
        return False


def test_camera_connection():
    print("\n🔹 اختبار اتصال كاميرا IP Webcam...")
    cap = cv2.VideoCapture(CAMERA_URL)
    if not cap.isOpened():
        print(" فشل الاتصال بالكاميرا. تحقق من العنوان والإعدادات.")
        return False
    print(" الاتصال بالكاميرا ناجح! اضغط Q للخروج.")
    while True:
        ret, frame = cap.read()
        if not ret:
            print("⚠️ فقد الاتصال بالفيديو.")
            break
        cv2.imshow("IP Webcam", frame)
        if cv2.waitKey(1) & 0xFF == ord('q'):
            break
    cap.release()
    cv2.destroyAllWindows()
    return True


# ===============================
# تشغيل الاختبارات
# ===============================
if __name__ == "__main__":
    gps_ok = test_gps_connection()
    cam_ok = test_camera_connection()

    if gps_ok and cam_ok:
        print("\n كل شيء جاهز! الاتصال بالهاتف والكاميرا ناجح.")
    else:
        print("\n تحقق من إعدادات الشبكة أو التطبيقات في الهاتف.")
```
